## Exercise 1: Tuple unpacking patterns

## Exercise 2: Tuple transformation

- Start from a tuple of integers.
- Create a new tuple with each value multiplied by 10.
- Print original and transformed tuples.

```python
# Write your solution here
original = (1, 2, 3, 4)
transformed = tuple(value * 10 for value in original)
print(original)
print(transformed)
```

## Exercise 3: Set-based analytics

- Given two sets of course names, print union, intersection, and both differences.
- Print symmetric difference and explain what it means in a comment.

```python
# Write your solution here
# Given two sets of course names
sem1_courses = {"Math", "Python", "Physics", "English"}
sem2_courses = {"Python", "Database", "Algorithms", "Physics"}

# Print union, intersection, and both differences
print("Union (all courses):", sem1_courses | sem2_courses)
print("Intersection (common courses):", sem1_courses & sem2_courses)
print("Difference (sem1 only):", sem1_courses - sem2_courses)
print("Difference (sem2 only):", sem2_courses - sem1_courses)

# Symmetric difference: elements in either set, but not in both
# In this context, it shows courses taken in exactly one semester
sym_diff = sem1_courses ^ sem2_courses
print("Symmetric difference (exclusive to one semester):", sym_diff)
```

## Exercise 4: Set cleanup task

- Create a set from a list containing repeated and mixed-case words.
- Normalize words to lowercase before inserting into set.
- Print final normalized unique set sorted alphabetically.

```python
# Write your solution here
words = ["Apple", "banana", "APPLE", "Cherry", "BANANA", "cherry", "Mango"]

normalized_set = {word.lower() for word in words}

sorted_words = sorted(normalized_set)

print("Original list:", words)
print("Normalized unique set:", normalized_set)
print("Sorted words:", sorted_words)
```

## Exercise 5: Set from a list comprehension

- Use a list comprehension to build the squares of numbers 1..20 where the square is even.
- Convert the list to a set to remove duplicates (if any).
- Print the resulting set and its length.
- Convert it to a sorted list and print.

```python
# 1. Create a list of even squares from 1 to 20
squares = [x**2 for x in range(1, 21) if x**2 % 2 == 0]

# 2. Convert the list to a set
square_set = set(squares)

# 3. Print the set and its length
print("Set:", square_set)
print("Length:", len(square_set))

# 4. Convert to a sorted list
sorted_squares = sorted(square_set)
print("Sorted list:", sorted_squares)
```
